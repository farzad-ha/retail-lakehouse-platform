```sql
-- Create Silver streaming table for cleaned and validated data
CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_categories

-- Define data quality expectations for the silver table. No NULLS or negative numbers
(
    CONSTRAINT valid_category_id
        EXPECT (category_id IS NOT NULL AND category_id > 0)
        ON VIOLATION DROP ROW,

    CONSTRAINT valid_category_name
        EXPECT (category_name IS NOT NULL)
        ON VIOLATION DROP ROW
)

-- Describe the purpose of the Silver table
COMMENT "Cleaned and validated categories table"

-- Define Silver table properties
TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false
)

AS

SELECT DISTINCT
    category_id,
    category_name,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_categories
;
```
