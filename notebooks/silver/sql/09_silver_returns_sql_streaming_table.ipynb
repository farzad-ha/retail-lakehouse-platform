```sql
-- Create silver table with cleaned, validated and deduplicated data plus meta columns

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_returns

(
    CONSTRAINT valid_return_id EXPECT (return_id IS NOT NULL AND return_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_order_item_id EXPECT (order_item_id IS NOT NULL AND order_item_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_refund EXPECT (refund IS NOT NULL AND refund >= 0) ON VIOLATION DROP ROW
)

COMMENT "Cleaned and validated returns data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    return_id,
    order_item_id,
    refund,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_returns
;
```
