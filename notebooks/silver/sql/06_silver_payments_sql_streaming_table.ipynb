```sql
-- Create silver table from bronze table along with cleaned, validated and deduplicated data plus metadata added from bronze table

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_payments

(
    CONSTRAINT valid_payment_id EXPECT (payment_id IS NOT NULL AND payment_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_order_id EXPECT (order_id IS NOT NULL AND order_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_amount EXPECT (amount IS NOT NULL AND amount > 0) ON VIOLATION DROP ROW
)

COMMENT "Cleaned and validated payments data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    payment_id,
    order_id,
    amount,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_payments
;
```
