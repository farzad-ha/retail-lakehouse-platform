```sql
-- Create silver table with cleaned, validated and deduplicated data plus meta columns

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_stores


(
    CONSTRAINT valid_store_id EXPECT (store_id IS NOT NULL AND store_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_city EXPECT (city IS NOT NULL) ON VIOLATION DROP ROW
)
COMMENT "Cleaned and validated stores data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- Prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    store_id,
    city,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_stores
;
```
