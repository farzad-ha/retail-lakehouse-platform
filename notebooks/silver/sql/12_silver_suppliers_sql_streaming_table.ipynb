```sql
-- Create silver table with cleaned, validated and deduplicated data plus meta columns

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_suppliers

(
    CONSTRAINT valid_supplier_id EXPECT (supplier_id IS NOT NULL AND supplier_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_country EXPECT (country IS NOT NULL) ON VIOLATION DROP ROW
)
COMMENT "Cleaned and validated suppliers data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- Prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    supplier_id,
    country,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_suppliers
;
```
