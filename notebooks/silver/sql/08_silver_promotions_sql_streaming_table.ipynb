```sql
-- Create silver table with cleaned, validated and deduplicated data plus meta columns

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_promotions

(
    CONSTRAINT valid_promotion_id EXPECT (promotion_id IS NOT NULL AND promotion_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_discount EXPECT (discount IS NOT NULL AND discount BETWEEN 0 AND 100) ON VIOLATION DROP ROW    
)

COMMENT "Cleaned and validated promotions data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    promotion_id,
    discount,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_promotions
;
```
