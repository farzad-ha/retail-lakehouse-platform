```sql
-- Create silver table with cleaned, validated and deduplicated data plus meta columns

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_shipments


(
    CONSTRAINT valid_shipment_id EXPECT (shipment_id IS NOT NULL AND shipment_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_order_id EXPECT (order_id IS NOT NULL AND order_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_status EXPECT (status IS NOT NULL AND status IN ('shipped', 'processing', 'delivered'))
)

COMMENT "Cleaned and validated shipment data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    shipment_id,
    order_id,
    status,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_shipments
;
```
