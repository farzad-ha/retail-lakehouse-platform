```sql
-- Create silver table from bronze table along with cleaned, validated and deduplicated data plus metadata added from bronze table

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_employees

(
    CONSTRAINT valid_employee_id EXPECT (employee_id IS NOT NULL AND employee_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_store_id EXPECT (store_id IS NOT NULL AND store_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_salary EXPECT (salary IS NOT NULL AND salary >= 0) ON VIOLATION DROP ROW
)
COMMENT "Cleaned and validated employees data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- Prevent accidental full refreshes of the table
)

AS

SELECT DISTINCT
    employee_id,
    store_id,
    salary,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_employees
;
```
