```sql
-- Create silver products table with cleaned, validated and deduplicated data plus meta columns

CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.silver.silver_products

(
    CONSTRAINT valid_product_id EXPECT (product_id IS NOT NULL AND product_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_category_id EXPECT (category_id IS NOT NULL AND category_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_supplier_id EXPECT (supplier_id IS NOT NULL AND supplier_id > 0) ON VIOLATION DROP ROW,

    CONSTRAINT valid_price EXPECT (price IS NOT NULL AND price >= 0) ON VIOLATION DROP ROW

)

COMMENT "Cleaned and validated products data"

TBLPROPERTIES (
    "quality" = "silver",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes on table
)

AS

SELECT DISTINCT
    product_id,
    category_id,
    supplier_id,
    price,
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM STREAM first_data_engineering_project.bronze.bronze_products
;
```
