```sql
-- NOTE: This notebook is for demonstration purposes and showcases Unity Catalog
-- access control patterns within this Databricks project.
--
-- Unity Catalog Access Control Examples
--
-- This notebook provides an example of common Unity Catalog permission patterns:
-- 1. Catalog-level permissions
-- 2. Schema-level permissions
-- 3. Table-level permissions
--
-- The groups used below are example groups to demonstrate security configuration.
-- In enterprise environments, these groups are typically managed through an identity provider,
-- such as Entra ID (Azure) or IAM/identity services in cloud environments.


-- Grant catalog access to data engineering users

GRANT USE CATALOG
ON CATALOG first_data_engineering_project
TO `data_engineers`;


-- Grant schema access to data engineering users

GRANT USE SCHEMA
ON SCHEMA first_data_engineering_project.bronze
TO `data_engineers`;

GRANT USE SCHEMA
ON SCHEMA first_data_engineering_project.silver
TO `data_engineers`;

GRANT USE SCHEMA
ON SCHEMA first_data_engineering_project.gold
TO `data_engineers`;
```
