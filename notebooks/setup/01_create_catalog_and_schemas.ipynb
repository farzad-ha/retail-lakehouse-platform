```sql
-- Create catalog
CREATE CATALOG IF NOT EXISTS first_data_engineering_project;


-- Create schemas
CREATE SCHEMA IF NOT EXISTS first_data_engineering_project.bronze;

CREATE SCHEMA IF NOT EXISTS first_data_engineering_project.silver;

CREATE SCHEMA IF NOT EXISTS first_data_engineering_project.gold;

CREATE SCHEMA IF NOT EXISTS first_data_engineering_project.monitoring;
```

```sql
-- Create landing schema to store incoming raw source files before ingestion
CREATE SCHEMA IF NOT EXISTS first_data_engineering_project.landing;
```

```sql
-- Create a managed volume for incoming source files
CREATE VOLUME IF NOT EXISTS first_data_engineering_project.landing.retail_files;
```

```sql
-- Create schema for pipeline operational metadata
CREATE SCHEMA IF NOT EXISTS first_data_engineering_project.pipeline_metadata;
```

```sql
-- Create volume to store Auto Loader metadata such as schemas and checkpoints
CREATE VOLUME IF NOT EXISTS first_data_engineering_project.pipeline_metadata.autoloader_metadata;
```

```sql
-- Create volumes in bronze schema to organize SQL and PySpark implementations
CREATE VOLUME IF NOT EXISTS first_data_engineering_project.bronze.sql;
CREATE VOLUME IF NOT EXISTS first_data_engineering_project.bronze.pyspark;
```
