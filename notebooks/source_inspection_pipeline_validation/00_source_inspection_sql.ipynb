```sql
-- Reusable notebook for inspecting source files before pipeline development
-- Used across Bronze, Silver, and Gold data workflows
```

```sql
-- Load CSV file for inspection
CREATE OR REPLACE TEMP VIEW -- source_data
AS
SELECT *
FROM read_files(
    -- '/Volumes/first_data_engineering_project/landing/retail_files/customers/',
    format => 'csv',
    header => true
);
```

```sql
-- Check sample records
SELECT *
FROM -- "source_data"
LIMIT 10;
```

```sql
-- Check total records
SELECT COUNT(*) AS total_rows
FROM -- "source_data";
```

```sql
-- Inspect columns and data types
DESCRIBE TABLE -- "source_data";
```

```sql
-- Check duplicate rows
SELECT 
    COUNT(*) AS total_rows, 
    COUNT(DISTINCT *) AS distinct_rows
FROM -- source_data;
```

```sql
-- Check missing customer IDs
SELECT
    SUM(CASE WHEN customer_id IS NULL THEN 1 ELSE 0 END) AS null_customer_id,
    SUM(CASE WHEN email IS NULL THEN 1 ELSE 0 END) AS null_email
FROM -- source_data;
```
