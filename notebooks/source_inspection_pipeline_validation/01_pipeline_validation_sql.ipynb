```sql
-- Reusable notebook for validating pipeline outputs
-- Used across Bronze, Silver, and Gold data layers
```

```sql
-- Check Bronze records
SELECT *
FROM first_data_engineering_project.bronze.bronze_customers
LIMIT 10;
```

```sql
-- Validate Bronze row count
SELECT
    COUNT(*) AS bronze_row_count
FROM first_data_engineering_project.bronze.bronze_customers;
```

```sql
-- Check Bronze metadata columns
SELECT
    ingestion_timestamp,
    source_file,
    file_modified_time
FROM first_data_engineering_project.bronze.bronze_customers
LIMIT 10;
```

```sql
-- Inspect Delta table details
DESCRIBE DETAIL first_data_engineering_project.bronze.bronze_customers;
```

```sql
-- Check latest ingestion time
SELECT
    MAX(ingestion_timestamp) AS latest_ingestion_time
FROM first_data_engineering_project.bronze.bronze_customers;
```

```sql
-- Check duplicate business keys
SELECT
    customer_id,
    COUNT(*) AS duplicate_count
FROM first_data_engineering_project.bronze.bronze_customers
GROUP BY customer_id
HAVING COUNT(*) > 1;
```
