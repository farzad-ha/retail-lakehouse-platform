```python
# Create struct schema for promotions csv file

from pyspark.sql.types import *

promotions_schema = StructType([
    StructField("promotion_id", IntegerType(), True),
    StructField("discount", IntegerType(), True)
])
```

```python
# autoload csv into dataframe with schema location defined

df = spark.readStream \
    .format("cloudFiles") \
    .option("cloudFiles.format", "csv") \
    .option("header", "true") \
    .option(
        "cloudFiles.schemaLocation", 
        "/Volumes/first_data_engineering_project/pipeline_metadata/autoloader_metadata/schemas/bronze/bronze_promotions/"
        ) \
    .schema(promotions_schema) \
    .load("/Volumes/first_data_engineering_project/landing/retail_files/promotions/")
```

```python
# add bronze layer metadata columns to the dataFrame for ingestion and lineage tracking

from pyspark.sql import functions as F
bronze_promotions = df \
    .withColumn("ingestion_timestamp", F.current_timestamp()) \
    .withColumn("source_file", F.col("_metadata.file_name")) \
    .withColumn("file_modified_time", F.col("_metadata.file_modification_time"))
```

```python
# create table with checkpoint location and add trigger

bronze_promotions.writeStream \
    .option(
        "checkpointLocation", 
        "/Volumes/first_data_engineering_project/pipeline_metadata/autoloader_metadata/checkpoints/bronze/bronze_promotions/"
        ) \
    .trigger(availableNow=True) \
    .toTable("first_data_engineering_project.bronze.bronze_promotions")
```
