```sql
-- Create Bronze streaming table with metadata columns and predefined schema
CREATE OR REFRESH STREAMING TABLE 
first_data_engineering_project.bronze.bronze_customers
COMMENT "ingest csv file customer from landing"
TBLPROPERTIES (
    "quality" = "bronze",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes of the Bronze ingestion table
    )
AS
SELECT
  *,
  current_timestamp() AS ingestion_timestamp,
  _metadata.file_name AS source_file,
  _metadata.file_modification_time AS file_modified_time
FROM STREAM read_files(
    '/Volumes/first_data_engineering_project/landing/retail_files/customers/',
    format => 'csv',
    header => true,
    schema => '
    customer_id	INT,
    city STRING, 
    signup_date	DATE
    '
);
```
