```sql

-- Create Bronze streaming table with metadata columns and predefined schema
CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.bronze.bronze_orders
COMMENT "ingest order csv files from landing"
TBLPROPERTIES (
    "quality" = "bronze",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes of the Bronze ingestion table
)
AS
SELECT 
    *,
    current_timestamp() AS ingestion_timestamp,
    _metadata.file_name AS source_file,
    _metadata.file_modification_time AS file_modified_time
FROM read_files(
    '/Volumes/first_data_engineering_project/landing/retail_files/orders/',
    format => 'csv',
    header => true,
    schema => '
    order_id INT, 
    customer_id INT, 
    store_id INT, 
    order_date DATE, 
    promotion_id INT
    '
);
```
