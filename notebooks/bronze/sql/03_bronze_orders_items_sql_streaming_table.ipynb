```sql
-- Create Bronze streaming table with metadata columns and predefined schema
CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.bronze.bronze_order_items
COMMENT "ingest csv files from order_items folder in landing"
TBLPROPERTIES (
    "quality" = "bronze",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes of the Bronze ingestion table
)
AS
SELECT
    *,
    current_timestamp() AS ingestion_timestamp,
    _metadata.file_name AS source_file,
    _metadata.file_modification_time AS file_modified_time
FROM STREAM read_files('/Volumes/first_data_engineering_project/landing/retail_files/order_items/',
format => 'csv',
header => true,
schema => '
    order_item_id INT,
    order_id INT,
    product_id INT,
    qty INT,
    price INT
    '
);
```
