```sql
CREATE OR REFRESH STREAMING TABLE first_data_engineering_project.bronze.bronze_products
COMMENT "ingest csv file product from landing"
TBLPROPERTIES(
    "quality" = "bronze",
    "pipelines.reset.allowed" = false -- prevent accidental full refreshes of the Bronze ingestion table
)
AS
SELECT
  *,
  current_timestamp() AS ingestion_timestamp,
  _metadata.file_name AS source_file,
  _metadata.file_modification_time AS file_modified_time
FROM STREAM read_files(
    '/Volumes/first_data_engineering_project/landing/retail_files/products/',
    format => 'csv',
    header => true,
    schema => '
    product_id INT,
    category_id INT,
    supplier_id INT,
    price INT
    '
);
```
